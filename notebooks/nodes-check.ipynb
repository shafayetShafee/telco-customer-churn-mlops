```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
```

```python
from telco_customer_churn_mlops.pipelines.ml_app.nodes_ml.threshold_tuning import tune_threshold
```

```python
scores = calib_model.predict_proba(X_calib)[:, 1]
```

```python
scores
```

```python
thresholds = np.linspace(0, 1, 101)
```

```python
results = []
pos_label = 1
zero_division = 0

for t in thresholds:
        y_pred = (scores >= t).astype(int)

        tn, fp, fn, tp = confusion_matrix(y_calib, y_pred).ravel()

        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0.0

        results.append({
            "threshold": float(t),
            "precision": precision_score(y_calib, y_pred, pos_label=pos_label, zero_division=zero_division),
            "recall": recall_score(y_calib, y_pred, pos_label=pos_label, zero_division=zero_division),
            "f1": f1_score(y_calib, y_pred, pos_label=pos_label, zero_division=zero_division),
            "tpr": tpr,
            "fpr": fpr,
            "accuracy": accuracy_score(y_calib, y_pred),
        })
```

```python
df = pd.DataFrame(results)
df.query("recall >= 0.80")
```
